```python
import numpy as np
import pandas as pd
```

### Distances (including the periodic boundary conditions)

```python
def calculate_distances(p1, p, side_length):
    """ p1: particle 1
        p: particles to calculate distance to 
        side length: side length of the 2d box
    """

    dx = np.abs(p1[0] - p[:,0])
    dy = np.abs(p1[1] - p[:,1])

    dx = np.where(dx > (side_length/2), side_length - dx, dx)
    dy = np.where(dy > (side_length/2), side_length - dy, dy)

    distances = np.sqrt(dx**2 + dy**2)
    return distances
```

### Set parameters

```python
L = 100             # side length of the square
N = 8*10**3         # number of simulation timesteps
dt = 1              # time step

n = 100             # number of particles
v = 1               # propulsion speed (same for all particles)

# flocking radius:
Rf_min = 1
Rf_max = 2

# noise level:
eta_min = 0
eta_max = 1

# number of samples:
num_samples = 400
```

### Saving the current parameters:

```python
parameters = pd.DataFrame()
```

```python
Rf_all = np.random.randint(Rf_min, Rf_max + 1, num_samples)     # generate random flocking radii
eta_all = np.random.uniform(eta_min, eta_max, num_samples)      # generate random noise levels
    
parameters['Rf'] = Rf_all
parameters['eta'] = eta_all

for sample in range(num_samples):
    print('sample:', sample)

    Rf = Rf_all[sample]
    eta = eta_all[sample]
    
    # initial positions and orientations
    x = np.zeros((N, n, 3))
    x[0,:,:2] = np.random.rand(n,2) * L                         # initial coordinates
    phi = np.random.rand(n) * 2 * np.pi                         # initial orientation

    ts = 0
    for i in range(1,N):
 
        # update positions:
        x[i,:,0] = (x[i-1,:,0] + v*dt*np.cos(phi)) % L          # update x-coordinates
        x[i,:,1] = (x[i-1,:,1] + v*dt*np.sin(phi)) % L          # update y-coordinates


        # update orientations:
        phi_updated = np.zeros_like(phi)
        for j in range(n):
            distances = calculate_distances(x[i,j], x[i], L)
            interact = distances < Rf
            phi_updated[j] = np.angle(np.sum(np.exp(phi[interact]*1j))) + eta * (np.random.rand()-0.5) * dt

        phi = phi_updated
```
